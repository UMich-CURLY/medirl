```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider
from PIL import Image

# Load and preprocess the image
uploaded_image_path = 'reward_temp.png'  # Update this to the actual file location
original_image = Image.open(uploaded_image_path).convert('L')  # Convert to grayscale

# Resize the image to 60x60
resized_image = original_image.resize((60, 60))
image_array = np.array(resized_image)

# Crop the middle 30x30 portion
height, width = image_array.shape
crop_size = 30
start_x = (width - crop_size) // 2
start_y = (height - crop_size) // 2
cropped_image = image_array[start_y:start_y + crop_size, start_x:start_x + crop_size]

# Function to apply bilateral filtering and display results
def apply_bilateral_filter(sigma_space, sigma_color):
    filtered_image = cv2.bilateralFilter(cropped_image, d=9, sigmaColor=sigma_color, sigmaSpace=sigma_space)
    plt.figure(figsize=(6, 6))
    plt.imshow(filtered_image, cmap='gray')
    plt.title(f'Bilateral Filtering\nSigma Space: {sigma_space}, Sigma Color: {sigma_color}')
    plt.axis('off')
    plt.show()

# Interactive sliders for sigma_space and sigma_color
interact(
    apply_bilateral_filter,
    sigma_space=IntSlider(min=1, max=50, step=1, value=15, description="Sigma Space"),
    sigma_color=IntSlider(min=1, max=150, step=1, value=75, description="Sigma Color"),
)
```
